## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
Task type: Ranking

The goal is to rank content pages by refresh opportunity. Instead of using a fixed manual review process,
the system uses observed content performance signals to create a priority order for SEO/content teams.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
Target: opportunity_score (proxy target)

The score will be created from observed signals:

Traffic decline (trend_direction, trend_pct)
Visibility (impressions_90d)
Engagement efficiency (ctr)
Content freshness (days_since_last_update)
Ranking opportunity (avg_position)

The score represents which pages may deserve review first, not which pages are guaranteed to improve.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
NDCG@K

It measures whether the highest-opportunity pages appear near the top of the ranked list.

A higher score means the ranking system better prioritizes pages that match the defined refresh opportunity signal.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
One row = one content page

Relevant dataframe columns:

content_id
impressions_90d
clicks_90d
ctr
avg_position
days_since_last_update
trend_direction
trend_pct
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
A fixed rule like:

"Refresh every page older than 365 days"

would miss important cases because content performance depends on multiple interacting factors.

For example:

An old page may still perform well.
A newer page may need attention because traffic is declining.
A high-impression page with low CTR may have more opportunity than a low-traffic page.

ML can learn patterns across multiple signals and support better prioritization.
```
